```python
import matplotlib.pyplot as plt
import numpy as np

T_scale = np.array([[3,0,0],
                    [0,1,0],
                    [0,0,1]])
T_shear = np.array([[1,2,0],
                    [0,1,0],
                    [0,0,1]])
T_translate = np.array([[1,0,0],
                        [0,1,0],
                        [1,3,1]])
T_combined = T_translate @ T_shear @ T_scale

P = np.array([2,-3,1])
P_transformed = T_combined @ P

plt.scatter(P[0], P[1], color = 'blue', s = 100, label = f'original P({P[0]},{P[1]})')
plt.scatter(P_transformed[0], P_transformed[1], color = 'red', s = 100, label = f'Transformed P({P_transformed[0]},{P_transformed[1]})')

plt.axhline(0, color = 'black')
plt.axvline(0, color = 'black')
plt.grid(True)
plt.xlim(-15,15)
plt.ylim(-15,15)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def bezier_curve(control_points, num_points = 100):
    n = len(control_points) - 1
    t = np.linspace(0,1,num_points)
    curve = np.zeros((num_points, 2))

    for i in range(num_points):
        for j in range(n + 1):
            binom = math.factorial(n) / (math.factorial(j) * math.factorial(n-j))
            bernstein = binom * (t[i]**j) * ((1 - t[i])**(n-j))
            curve[i] += bernstein * control_points[j]
    return curve
```

```python
control_points = np.array([[4,3],
                           [0,1],
                           [2,-1]])
curve = bezier_curve(control_points)
plt.figure(figsize = (8,6))
plt.plot(curve[:,0], curve[:,1], 'r-', linewidth = 3, label = 'bezier curve')
plt.plot(control_points[:,0], control_points[:,1], 'bo--', linewidth = 1, label = 'Control points')
plt.plot(control_points[:,0], control_points[:,1], 'bo', markersize = 8)
plt.grid()
plt.legend()
plt.title("Bezier curve: b0, b1, b2")
plt.show()
```
