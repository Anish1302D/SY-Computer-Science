```python
#set b q1

class person:
    def __init__(self, name, age):
        self._name = name
        self._age = age
        
    @property
    def name(self):
        return self._name
    
    @property
    def age(self):
        return self._age
    
    def display(self):
        print("name : ", self._name)
        print("age : ", self._age)
        
class student(person):
    def __init__(self, name, age, rollno, marks):
        super().__init__(name, age)
        self._rollno = rollno
        self._marks = marks
        
    @property
    def rollno(self):
        return self._rollno
    
    @property
    def marks(self):
        return self._marks 
    
    def display(self):
        super().display()
        print("rollnumber : ", self._rollno)
        print("marks : ", self._marks)
        
name = input("enter name : ")        
age = int(input("enter age : "))
roll = int(input("enter roll number : "))
marks = float(input("enter marks : "))

s = student(name, age, roll, marks)
print("\n student details :  ")
s.display()
```

```python
class vehicle:
    def start(self):
        print("starting vehicle")
        
class car(vehicle):
    def start(self):
        print("starting car")
        
choice = input("enter type (vehicle/car) : ").lower()

if choice == "vehicle":
    v = vehicle()
    v.start()
    
elif choice == "car":
    c = car()
    c.start()
    
else : 
    print("invalid choice")
```

```python
class shape:
    def area(self):
        print("area method in shape class")
        
class rectangle(shape):
    def __init__(self,length, width):
        self.length = length
        self.width = width
    
    def area(self):
        rec_area = self.length * self.width
        print("area of rectangle : ",rec_area)
        
class triangle(shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        tri_area = 0.5 * self.base * self.height
        print("area of triangle = ",tri_area)
        
length = float(input("enter length of rectangle : "))
width = float(input("enter width of rectangle: "))
 
base = float(input("enter base of triangle : "))
height = float(input("enter height of triangle : "))

rec = rectangle(length, width)
tri = triangle(base, height)

shapes = [rec , tri]

for s in shapes:
    s.area()
```

```python
from abc import ABC, abstractmethod

class appliance(ABC):
    def operate(self):
        pass
    
class washingmachine(appliance):
    def __init__(self, mode):
        self.mode = mode
        
    def operate(self):
        print("washing machine is operating in" , self.mode,"mode")
        
class microwave(appliance):
    def __init__(self, time):
        self.time = time
        
    def operate(self):
        print("microwave is heating for", self.time, "minutes")

wm_mode = input("enter washing machine mode : ")
mw_time = input("enter microwave heating time (minutes) : ")

wm = washingmachine(wm_mode)
mw = microwave(mw_time)

wm.operate()
mw.operate()
```

```python
class teacher:
    def __init__(self, name, subject):
        self._name = name
        self._subject = subject
        
    @property
    def name(self):
        return self._name
    
    @property
    def subject(self):
        return self._subject
    
    def display_info(self):
        print("teacher name: ", self._name)
        print("subject name: ", self._subject)
        
class researcher:
    def __init__(self, field, publications):
        self._field = field
        self._publications = publications
        
    @property
    def field(self):
        return self._field
    
    @property
    def publications(self):
        return self._publications
    
    def display_info(self):
        print("research field: ",self._field)
        print("publication : ",self._publications)
        
class professor(teacher , researcher):
    def __init__ (self, name, subject, field, publications):
        teacher.__init__ (self, name, subject)
        researcher.__init__(self, field, publications)

name = input("enter a name: ")
subject = input("enter a subject : ")
field = input("enter a research field : ")
pub = input("enter number of publications : ")

p = professor(name, subject, field, pub)

print("\n calling display_info()")
p.display_info()

print("\n method resolution order (MRO): ")
print(professor.__mro__)
```

```python
from abc import ABC, abstractmethod

class payment(ABC):
    def __init__(self, amount):
        self._amount = amount
        
    def amount(self):
        return self._amount
    
    def make_payment(self):
        pass
    
class creditcardpayment(payment):
    def make_payment(self):
        print(f"proccessing credit card payment of {amount}")
        
class upipayment(payment):
    def make_payment(self):
        print(f"proccessing upi paymentbof {amount}")
        
class netbankingpayment(payment):
    def make_payment(self):
        print(f" proccessing net banking payment of {amount}")
        
amount = float(input("enter amount: "))
method = input("enter payment method(credit / upi / netbanking:")

if method.lower()== "credit":
    p = creditcardpayment(amount)
    
elif method.lower() == "upi":
    p = upipayment(amount)
    
elif method.lower() == "netbanking"    :
    p = netbankingpayment(amount)
    
else:
    print("invalid payment method")
    exit()

p.make_payment()    
```
