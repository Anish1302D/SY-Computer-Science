```python
class student:
    def __init__(self, name, rollno, marks):
        self.name = name
        self.rollno = rollno
        self.marks = marks
        
    def display_details(self):
        print("-"*20)
        print(f"Name :{self.name}")
        print(f"Rollno :{self.rollno}")
        print(f"Marks :{self.marks}")
        
#accepting for two students
students = []
for i in range(1, 3):
    print(f"\n Enter details for student{i}: ")
    name = input("Enter name: ")
    roll = input("Enter roll number: ")
    marks = input("Enter marks: ")
    students.append(student(name, roll, marks))
    
#displaying details
print("\n -- Student records --")
for s in students:
    s.display_details()
```

```python
class calculator:
    def __init__(self, a, b):
        self.num1 = a
        self.num2 = b
        
    def add(self):
        return self.num1 + self.num2
    
    def sub(self):
        return self.num1 - self.num2
    
    def product(self):
        return self.num1 * self.num2
    
    def divide(self):
        if self.num2 != 0:
            return self.num1 / self.num2
        else:
            return "Error! Division by Zero"
        
print("\n Simple Calculator")
n1 = float(input("Enter First Number"))
n2 = float(input("Enter Second Number"))

calc = calculator(n1, n2)

print(f"\n Results for {n1} and {n2}: ")
print (f"Addition: {calc.add()}")
print (f"Subtraction: {calc.sub()}")
print (f"Multiplication: {calc.product()}")
print (f"Division: {calc.divide()}")
```

```python
class employee:
    def __init__(self, emp_id, name, base_salary):
        self.emp_id = emp_id
        self.name = name
        self.base_salary = base_salary
        
    def display_info(self):
        print(f"[{self.emp_id}] Name: {self.name} | Base Salary: ${self.base_salary}")
             
    def work(self):
              return"Performing General Duties"
              
              
class developer(employee):
              def __init__(self, emp_id, name, base_salary, language):
                   super ().__init__(emp_id, name, base_salary)
                   self.language = language
              
              def work(self):
                  return f"Writing Code in {self.language}"
              
class tester(employee):
              def __init__(self, emp_id, name, base_salary, tool):
                super().__init__(emp_id, name, base_salary)
                self.tool = tool
              
              def work(self):
                  return f"Testing Software sing {self.tool}"
              
class manager(employee):
              def __init__(self, emp_id, name, base_salary, team_size):
                   super().__init__(emp_id, name, base_salary)
                   self.team_size = team_size
              
              def work(self):
                   return f"Managing a Team of {self.team_size} Employees"
              
dev = developer(101, "Alice", 80000, "Python")
test = tester(102, "Bob", 65000, "Selenium")
mgr = manager(103, "Charlie", 95000, 10)
              
employees = [dev, test, mgr]
              
print("Employees Task List: ")

for emp in employees:
    emp.display_info()
    print(f"Task: {emp.work()}\n")          
              
```

```python
class atm:
    def __init__ (self,pin):
        self.__pin = pin
        
    def validate_pin(self, entered_pin):
        """Public method to check the pin"""
        if entered_pin == self.__pin:
            print("Access Granted! Welcome to your Account")
            return True
        else:
            print("Access Denied! Incorrect Pin")
            return False
        
my_atm = atm("1302")
user_input = input("Please Enter your 4 Digit Pin: ")
my_atm.validate_pin(user_input)
```

```python
class user:
    """Base Class for all System Users"""
    def __init__(self,name,phone):
        self.name = name
        self.phone = phone
        
    def display_profile(self):
        print(f"User: {self.name} | Phone: {self.phone}")
        
class customer(user):
    """Derived Class representing a person placing orders"""
    def __init__(self, name, phone, address):
        super().__init__(name, phone)
        self.address = address
        
    def show_customer_info(self):
        self.display_profile()
        print(f"Delivery Address: {self.address}")
        
class deliveryperson(user):
    """Derived Class representing the person delivering orders"""
    def __init__(self, name, phone, vehicle_number):
        super().__init__(name, phone)
        self.vehicle_number = vehicle_number
        
    def show_delivery_info(self):
        super().display_profile()
        print(f"Vehicle No.: {self.vehicle_number}")
        
print("  Creating Customer  ")
cust = customer("Arjun","8493846573","123 Green Avenue Pune" )
cust.show_customer_info()

print("  Creating Delivery Person  ")
rider  = deliveryperson("Suresh","4747859484","Tilak Road Pune" )
rider.show_delivery_info()
                
```

```python
class university:
    def university_info(self, uni_name):
        self.uni_name = uni_name
        print(f"University: {self.uni_name}")
        
class college(university):
    def college_info(self, college_name):
        self.college_name = college_name
        print(f"College: {self.college_name}")
    
class student(college):
    def student_info(self, student_name):
        self.student_name = student_name
        print(f"Student Name: {self.student_name}")
        
s1 = student()
print("Accessing the methods accross the chain ")

s1.university_info(" SPPU ")
s1.college_info(" SP ")
s1.student_info(" Anish ")
```

```python
from abc import ABC, abstractmethod

class Database(ABC):
    def __init__(self):
        self.connection = None
        self.connect()
        
    @abstractmethod
    def connect(self):
        """Establish database connection"""
        pass
    
    @abstractmethod
    def close(self):
        """Close database connection"""
        pass
        
    def __del__(self):
        #Destructor: Called when object is destroyed
        if self.connection:
            self.close()
                
class MySQLDatabase(Database):
    def connect(self):
        #Simulating a database connection 
        self.connection = "MySQL Connection Established"
        print(self.connection)
        
    def close(self):
        print("MySQL Connection Closed")
        self.connection = None
        
#Using the database
db = MySQLDatabase()

#When program ends or object is deleted, _del_() is called automatically
```

```python
from abc import ABC, abstractmethod

#Abstract base calss
class Shape(ABC):
    
    @abstractmethod
    def draw(self):
        pass
    
#Concrete class circle
class Circle(Shape):
    def draw(self):
        print("Drawing a circle")
        
#Concrete class Rectangle
class Rectangle(Shape):
    def draw(self):
        print("Drawing a Rectangle")

c = Circle()
r = Rectangle()

c.draw()
r.draw()
        
```
