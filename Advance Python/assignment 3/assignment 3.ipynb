```python
class NegativeNumberError(Exception):
    pass

def checknum(num):
    if num < 0:
        raise NegativeNumberError("Negative number not allowed")
    else:
        print("valid number: ",num)
        
try:
    n = int(input("enter a number: "))
    checknum(n)
except NegativeNumberError as e:
     print("error:",e)
```

```python
try:
    num = int(input("enter a number: "))
    result = 10/num
except ZeroDivisionError:
    print("cannot divide by zero")
except ValueError:
    print("invalid input")
else:
    print("result: ",result)
finally:
    print("execution completed")
```

```python
try:
    with open("number.txt","r") as file:
        for line in file:
            try:
                num = int(line.strip())
                print("valid number: ",num)
            except ValueError:
                print("invalid data skipped: ",line.strip())
except FileNotFoundError:
    print("file not found")
```

```python
class LengthExceededError(Exception):
    pass

try:
    text = input("enter string: ")
    if len(text) > 10:
        raise LengthExceededError("string too long")
    print("valid string")
except LengthExceededError as e:
    print("error: ", e)
```

```python
balance = 5000

try:
    amount = int(input("enter withdrawal amount: "))
    if amount > balance:
        raise ValueError("insufficient balance")
    balance -= amount
    print("transaction succesfull",balance)
except ValueError as e:
    print("error : ",e)
```

```python
try:
    num = int(input("enter a number: "))
    if num < 0:
        raise ValueError("negative number not allowed")
except ValueError as e:
    print("error :" ,e)
else:
    print("valid number entered: ",num)
finally:
    print("program ended")
```

```python
class InvalidAge(Exception):
    pass

try:
    age = int(input("enter age : "))
    if age < 18:
        raise InvalidAge("age must be 18 or above")
    print("eligible")
except InvalidAge as e:
    print("error: ",e)
except ValueError:
    print("invalid input")
```
