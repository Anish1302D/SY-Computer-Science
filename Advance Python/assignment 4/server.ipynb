```python
import socket

server_socket = socket.socket()
server_socket.bind(("localhost", 53000))

server_socket.listen(1)
print("Server started")

conn, addr = server_socket.accept()

num = int(conn.recv(1024).decode())

flag = True
if num <= 1:
    flag = False
else:
    for i in range(2, num):
        if num % i == 0:
            flag = False
            break

if flag:
    print("Number is Prime")
else:
    print("Number is NOT Prime")
```

```python
import socket

server_socket = socket.socket()
server_socket.bind(("localhost", 54000))

server_socket.listen(1)
print("Server started")

conn, addr = server_socket.accept()

while True:
    msg = conn.recv(1024).decode()
    
    if not msg:
        break
        
    if msg == "quit":
        print("Client ended communication")
        break
        
    print("Message:", msg)

conn.close()
server_socket.close()
```

```python
import socket

server_socket = socket.socket()
server_socket.bind(("localhost", 55000))

server_socket.listen(1)
print("Server started")

conn, addr = server_socket.accept()
print("Connection from", addr)

with open("received_file.bin", "wb") as file:
    while True:
        data = conn.recv(1024)
        if not data:
            break
        file.write(data)

print("File received successfully")

conn.close()
server_socket.close()
```
